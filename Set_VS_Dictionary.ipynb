# dictionary

```python
Dictionart1 = {"name": "omar","age":25,"Country":"Alex"}
print((Dictionart1))
```

```python
Dictionart = {"num1":1,"num2":2}
print(Dictionart["num1"])
```

```python
Dictionart = {"num1":1,"num2":2}
Dictionart["num1"] = 8
print(Dictionart)
```

```python
dictionary1 = {"name":"omar",
               "age":20,
               "dict2":{"name":"Ali",
                        "age":20,
                        "Faculty":"CS"}}
dictionary1["dict2"]["Faculty"] = "AI"
print(dictionary1)
```

```python
List_To_Dict = [("a",2),("b",3),("C",4)]
dict1 = dict(List_To_Dict)
print(dict1)
```
