# List

## Print Type of List

```python
list_one = [1, 2, 3, 4, 5]
print(type(list_one))
```

## List is a Data Ordered 

```python
list_one = [1, 2, 3,[ 4, 5]]
print(list_one[3][1])
```

## List is Changeable Data

```python
list_one = [1, 2, 3,[ 4, 5]]
list_one[1] = "Python"
print(list_one)
```

## List Allow Duplicate Data
## List Can be nested among all

```python
list_one = [1, 2, 3,[ 4, 5],[4,5,7,8,9,[1,2,3,4,5]]]
print(list_one)
```

## Convert any datatype to list using List() function

```python
name = ("omar")
list_one = list(name)
print(list_one)
```

## Add Items to a List

### append()
Takes an item and adds it to the end of the list.

### insert()
Takes an index first, then an item, and adds the item at that specific index.

### extend()
Takes an iterable and adds its items to the list.

```python
list_one = [1,2,3,4,5,6,"Dev",True,False,[1,2,3,4,["ML","AI"]]]
list_one.append("Python")
print(list_one)
```

```python
list_one = [1,2,3,4,5,6,"Dev",True,False,[1,2,3,4,["ML","AI"]]]
list_one.insert(0, "Python")
print(list_one)
```

```python
list_one = [1,2,3,4,5,6,"Dev",True,False,[1,2,3,4,["ML","AI"]]]
list_one.extend("Python")
print(list_one)
```

## Remove Items from a List

### remove()
Removes the first occurrence of a specific item from the list.

### pop()
Removes and returns an item at a specific index. If no index is given, it removes the last item.

### del
Deletes an item or multiple items from the list using their index.

```python
list_one = [1,2,3,4,5,6,"Dev",True,False,[1,2,3,4,["ML","AI"]]]
list_one.remove(3)
print(list_one)
```

```python
list_one = [1,2,3,4,5,6,"Dev",True,False,[1,2,3,4,["ML","AI"]]]
list_one.pop(9)
print(list_one)
```

```python
list_one = [1,2,3,4,5,6,"Dev",True,False,[1,2,3,4,["ML","AI"]]]

del list_one [1]
print(list_one)
```

## Get Input from the User

### input()
Takes input from the user and returns it as a string.

```python
name =  input("Enter your name: ")
age =  input("Enter your age: ")
print(f"Hello {name} your age is {age}")
```

## Other List Methods

### index()
Returns the index of the first occurrence of a specific item.

### count()
Returns the number of times an item appears in the list.

### sort()
Sorts the items in the list.

### reverse()
Reverses the order of the items in the list.

### clear()
Removes all items from the list.

### del
Deletes a list from the memory.

```python
list_one = [1,2,3,4,5,6,"Dev",True,False,[1,2,3,4,["ML","AI"]]]
index = list_one.index("Dev")
print(index)
```

```python
list_one = [1,2,3,4,5,6,"Dev",True,False,[1,2,3,4,["ML","AI"]],[1,2,3,4,["ML","AI"]],[1,2,3,4,["ML","AI"]]]
count = list_one.count([1,2,3,4,["ML","AI"]])
print(count)
```

```python
list_one = [1,2,3,4,20,5,3,1]
list_one.sort()
print(list_one)
```

```python
list_one = [1,2,3,4,20,5,3,1]
list_one.reverse()
print(list_one)
```

```python
list_one = [1,2,3,4,20,5,3,1]
list_one.clear()
print(list_one)
```

```python
list_one = [1, 2, 3, 4, 5,1,2,3,4]
del list_one
print(list_one)
```

# Tuple


## print type of tuple

```python
tuple_one = (1,2,3,4,"Dev",False)
print(type(tuple_one))
```

## Tuple is ordered Data

```python
tuple_one = (1,2,3,4,"Dev",False)
print(tuple_one[2])
```

## Tuple is unchangeable Data

```python
tuple_one = (1,2,3,4,"Dev",False)
tuple_one[2] = "Python"
print(tuple_one)
```

```python
list_one = list(tuple_one)
print(list_one)
```

```python
list_one[2]= "Python"
print(list_one)
```

```python
tuple_one = tuple(list_one)
print(tuple_one)
```

## Tuple Can be nested among all


```python
tuple_one = (1,2,3,(4,"Dev",False))
print(tuple_one)
```

## Convert any datatype to tuple using tuple() function


```python
name = ["omar","ali"]
tuple_one = tuple(name)
print(tuple_one)
```

```python
tuple_one = (1,2,3,4,"Dev",False)
tuple2 = list(tuple_one)
tuple2.insert(2,"Python")
print(tuple2)
```
